```python
import pandas as pd

df = pd.read_csv("sample_trajectories.csv")
df.head()
```

```python
from detailviews import DetailViewsWidget

widget = DetailViewsWidget.from_dataframe(
    df, x="x", y="y", trajectory="episode", order="step"
)
widget
```

**What to try:** lasso a few points (left-drag) — inset detail views and
cluster contours appear; `progress` and `speed` show up as feature columns in
the insets and in the color mapping. Any extra column of your DataFrame is
picked up the same way.
