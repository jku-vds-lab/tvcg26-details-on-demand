This notebook rehearses the exact experience a user gets **after** the package
is published to PyPI: install by name, load a CSV with pandas, get the widget.

Today it installs from the locally built wheel instead of PyPI — same resolver, same
name, nothing public (dependencies still come from PyPI; `detailviews`
itself resolves from the local `dist/` folder because the name has never
been uploaded). Build the wheel first (once):

```bash
cd packages/widget
npm run build:widget
python -m pip install build

```

That leaves `detailviews-0.1.0-py3-none-any.whl` in `packages/widget/dist/`.
After the real PyPI publish, the install cell below loses its two local flags
and becomes plain `%pip install detailviews` — everything else stays identical.

````python
# Post-publish this cell is just:  %pip install detailviews
%pip install detailviews --find-links ../dist
````

If `detailviews` was already imported in this kernel before installing,

````python
import pandas as pd

df = pd.read_csv("sample_trajectories.csv")
df.head()
````

````python
from detailviews import DetailViewsWidget

widget = DetailViewsWidget.from_dataframe(
    df, x="x", y="y", trajectory="episode", order="step"
)
widget
````
