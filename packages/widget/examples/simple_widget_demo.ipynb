# Trajectory widget demo (simple format)

Renders the rl_trajectories visualization as a Jupyter widget with a dataset passed in from Python.

**Setup (once, from the repo root — use the same Python your notebook kernel runs on):**

```bash
python packages/widget/examples/setup_widget.py
```

The script builds the gitignored JS bundle (`npm ci` + `npm run build:widget`) and
editable-installs `detailviews` into that Python's environment, then verifies the import.
Running it with the kernel's interpreter is what guarantees `import detailviews` works here —
if this notebook still can't find the module, the kernel is a different environment than
the one you ran the script with (check `import sys; sys.executable`).

The dataset is the *simple format*: rows with projected `x`/`y` columns, arbitrary feature
columns, and optionally a trajectory-id column and a within-trajectory order column.
The kNN graph, trajectory splines, and HDBSCAN clustering are computed inside the widget,
off the UI thread.

````python
import math
import random

# A small synthetic dataset: 8 spiral trajectories x 30 steps.
rows = []
for traj in range(8):
    rng = random.Random(traj)
    angle0 = rng.random() * 2 * math.pi
    cx, cy = rng.uniform(-6, 6), rng.uniform(-6, 6)
    for step in range(30):
        t = step / 29
        r = 1 + 4 * t
        rows.append({
            "x": cx + r * math.cos(angle0 + 3 * t),
            "y": cy + r * math.sin(angle0 + 3 * t),
            "episode": traj,
            "step": step,
            "progress": round(t, 3),
        })
len(rows)
````

````python
from detailviews import DetailViewsWidget

widget = DetailViewsWidget(
    data=rows,
    columnMapping={"x": "x", "y": "y", "trajectory": "episode", "order": "step"},
    datasetType="default",
)
widget
````

With pandas, the equivalent is:

```python
import pandas as pd
df = pd.DataFrame(rows)
widget = DetailViewsWidget.from_dataframe(
    df, x="x", y="y", trajectory="episode", order="step"
)
widget
```
